```python
from pyspark.sql.functions import col, sha2, to_date

# 1. Standardize Dates & Handle Privacy
# We'll format Admission and Discharge dates instead of BirthDate
refined_patients_df = spark.table("silver_patients") \
    .withColumn("AdmissionDate", to_date(col("Admission_Date"), "yyyy-MM-dd")) \
    .withColumn("DischargeDate", to_date(col("Discharge_Date"), "yyyy-MM-dd")) \
    .withColumn("Patient_Hash", sha2(col("Patient_ID"), 256)) 

# 2. Select the columns that actually exist in your data
final_patients_df = refined_patients_df.select(
    "Patient_Hash", 
    "Age", 
    "Gender", 
    "AdmissionDate", 
    "DischargeDate",
    "Diagnosis", 
    "Treatment_Cost",
    "Outcome"
)

# 3. Overwrite the Silver table
final_patients_df.write.format("delta") \
    .mode("overwrite") \
    .option("overwriteSchema", "true") \
    .saveAsTable("silver_patients")

print("Silver Layer Refined!")
```

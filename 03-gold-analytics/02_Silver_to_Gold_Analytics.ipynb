```python
from pyspark.sql.functions import col, avg, sum, count, round

# 1. Load the Silver data
silver_df = spark.table("silver_patients")

# 2. Create the Gold Aggregate: Cost and Outcome Analysis
gold_performance_df = silver_df.groupBy("Diagnosis", "Outcome") \
    .agg(
        count("Patient_Hash").alias("Total_Patients"),
        round(avg("Treatment_Cost"), 2).alias("Avg_Cost"),
        round(sum("Treatment_Cost"), 2).alias("Total_Revenue"),
        round(avg("Age"), 1).alias("Avg_Patient_Age")
    ) \
    .orderBy(col("Total_Revenue").desc())

# 3. Save as a Gold Table
gold_performance_df.write.format("delta") \
    .mode("overwrite") \
    .saveAsTable("gold_clinical_performance")

print("Gold Layer Created: Clinical Performance Summary is ready.")
```

```python
# FILE: 01_Bronze_Ingestion
# Objective: Move raw files to the initial Delta Lake landing zone.

# 1. Read the raw CSV from the 'Files' section
# Replace 'patients.csv' with your actual filename from Kaggle
raw_data = spark.read.format("csv") \
    .option("header", "true") \
    .option("inferSchema", "true") \
    .load("Files/Bronze/patients.csv")

# 2. Save it as a Raw Table (Bronze)
# We do NOT clean it yet. This is our 'Source of Truth'.
raw_data.write.format("delta") \
    .mode("overwrite") \
    .saveAsTable("bronze_patients")

print("Bronze Ingestion Complete: Raw data is now a Delta Table.")
```
